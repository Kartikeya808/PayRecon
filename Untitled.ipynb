```python
SYMPTOMS = [
    "frequent_urination",
    "excessive_thirst",
    "unexplained_weight_loss",
    "fatigue",
]

SYMPTOM_LABELS = {
    "frequent_urination": "frequent urination",
    "excessive_thirst": "excessive thirst",
    "unexplained_weight_loss": "unexplained weight loss",
    "fatigue": "fatigue",
}

def make_patient(name, symptoms, fasting_glucose=None, random_glucose=None):
    """Create a patient record. Unknown symptoms should be None."""
    patient_symptoms = {key: symptoms.get(key, None) for key in SYMPTOMS}
    return {
        "name": name,
        "symptoms": patient_symptoms,
        "fasting_glucose": fasting_glucose,
        "random_glucose": random_glucose,
    }

print("Patient input structure ready.")
```

```python
KNOWLEDGE_BASE = [
    {
        "id": "R1",
        "description": "Frequent urination with excessive thirst",
        "condition": lambda p: (
            p["symptoms"]["frequent_urination"] is True
            and p["symptoms"]["excessive_thirst"] is True
        ),
        "conclusion": "screening",
        "recommendation": "Arrange diabetes screening with a healthcare professional.",
        "explanation": "Frequent urination and excessive thirst are symptoms associated with possible diabetes."
    },
    {
        "id": "R2",
        "description": "Unexplained weight loss with fatigue",
        "condition": lambda p: (
            p["symptoms"]["unexplained_weight_loss"] is True
            and p["symptoms"]["fatigue"] is True
        ),
        "conclusion": "medical_evaluation",
        "recommendation": "Arrange a medical evaluation for these symptoms.",
        "explanation": "Unexplained weight loss together with fatigue warrants clinical assessment."
    },
    {
        "id": "R3",
        "description": "Excessive thirst with fatigue",
        "condition": lambda p: (
            p["symptoms"]["excessive_thirst"] is True
            and p["symptoms"]["fatigue"] is True
        ),
        "conclusion": "screening",
        "recommendation": "Discuss diabetes screening with a healthcare professional.",
        "explanation": "Excessive thirst and fatigue can occur with diabetes but are not specific to it."
    },
    {
        "id": "R4",
        "description": "Fasting glucose at or above 126 mg/dL",
        "condition": lambda p: (
            p["fasting_glucose"] is not None
            and p["fasting_glucose"] >= 126
        ),
        "conclusion": "prompt_evaluation",
        "recommendation": "Seek prompt clinical evaluation of the fasting glucose result.",
        "explanation": "A fasting glucose result at or above 126 mg/dL is in the diabetes-range threshold and requires clinical interpretation and usually confirmation."
    },
    {
        "id": "R5",
        "description": "Random glucose at or above 200 mg/dL with classic symptoms",
        "condition": lambda p: (
            p["random_glucose"] is not None
            and p["random_glucose"] >= 200
            and (
                p["symptoms"]["frequent_urination"] is True
                or p["symptoms"]["excessive_thirst"] is True
                or p["symptoms"]["unexplained_weight_loss"] is True
            )
        ),
        "conclusion": "prompt_evaluation",
        "recommendation": "Seek prompt clinical evaluation of the random glucose result and symptoms.",
        "explanation": "A random glucose result at or above 200 mg/dL with classic symptoms warrants prompt clinical assessment; this program does not make a diagnosis."
    },
    {
        "id": "R6",
        "description": "Three or more diabetes-associated symptoms",
        "condition": lambda p: sum(
            p["symptoms"][symptom] is True for symptom in SYMPTOMS
        ) >= 3,
        "conclusion": "screening",
        "recommendation": "Arrange diabetes screening and discuss the symptoms with a healthcare professional.",
        "explanation": "Several associated symptoms occurring together support further screening, although symptoms alone cannot establish a diagnosis."
    },
]
print(f"Loaded {len(KNOWLEDGE_BASE)} production rules.")
```

```python
def get_missing_information(patient):
    missing = []

    for symptom in SYMPTOMS:
        if patient["symptoms"][symptom] is None:
            missing.append(f"Whether the patient has {SYMPTOM_LABELS[symptom]}")

    if patient["fasting_glucose"] is None:
        missing.append("Fasting blood glucose measurement")
    if patient["random_glucose"] is None:
        missing.append("Random blood glucose measurement")

    return missing


def build_working_memory(patient):
    """Represent known patient information as facts for inference."""
    facts = set()

    for symptom, present in patient["symptoms"].items():
        if present is True:
            facts.add(f"symptom:{symptom}")
        elif present is False:
            facts.add(f"absent:{symptom}")

    if patient["fasting_glucose"] is not None:
        facts.add(f"fasting_glucose:{patient['fasting_glucose']}")

    if patient["random_glucose"] is not None:
        facts.add(f"random_glucose:{patient['random_glucose']}")

    return facts
```

```python
def forward_chain(patient, knowledge_base=KNOWLEDGE_BASE):
    working_memory = build_working_memory(patient)
    activated_rules = []
    fired_rule_ids = set()
    recommendations = []

    changed = True
    while changed:
        changed = False

        for rule in knowledge_base:
            if rule["id"] not in fired_rule_ids and rule["condition"](patient):
                fired_rule_ids.add(rule["id"])
                activated_rules.append(rule)
                recommendations.append({
                    "rule_id": rule["id"],
                    "recommendation": rule["recommendation"],
                    "explanation": rule["explanation"],
                })

                # Record the rule's conclusion in working memory.
                working_memory.add(f"conclusion:{rule['conclusion']}")
                changed = True

    return {
        "working_memory": working_memory,
        "activated_rules": activated_rules,
        "recommendations": recommendations,
        "missing_information": get_missing_information(patient),
    }

print("Forward-chaining inference engine defined.")
```

```python
def display_result(patient):
    result = forward_chain(patient)

    print("=" * 72)
    print(f"PATIENT: {patient['name']}")
    print("-" * 72)

    print("Symptoms:")
    for symptom in SYMPTOMS:
        value = patient["symptoms"][symptom]
        status = "Present" if value is True else "Absent" if value is False else "Unknown"
        print(f"  - {SYMPTOM_LABELS[symptom]}: {status}")

    print(f"Fasting glucose: {patient['fasting_glucose'] if patient['fasting_glucose'] is not None else 'Not provided'} mg/dL")
    print(f"Random glucose: {patient['random_glucose'] if patient['random_glucose'] is not None else 'Not provided'} mg/dL")

    print("\nActivated rules:")
    if result["activated_rules"]:
        for rule in result["activated_rules"]:
            print(f"  {rule['id']}: {rule['description']}")
    else:
        print("  No production rules were activated by the available information.")

    print("\nRecommendations and explanations:")
    if result["recommendations"]:
        for item in result["recommendations"]:
            print(f"  [{item['rule_id']}] {item['recommendation']}")
            print(f"      Reason: {item['explanation']}")
    else:
        print("  No rule-based recommendation was triggered.")
        print("  This does not rule out diabetes. Seek professional advice if symptoms or concerns persist.")

    print("\nMissing information:")
    if result["missing_information"]:
        for item in result["missing_information"]:
            print(f"  - {item}")
    else:
        print("  None of the tracked fields are missing.")

    print("\nDisclaimer: This is an educational screening prototype, not a diagnosis.")
    return result
```

```python
patient_1 = make_patient(
    name="Scenario 1",
    symptoms={
        "frequent_urination": True,
        "excessive_thirst": True,
        "unexplained_weight_loss": True,
        "fatigue": True,
    },
    fasting_glucose=132,
    random_glucose= None,
)

result_1 = display_result(patient_1)
```

```python
patient_2 = make_patient(
    name="Scenario 2",
    symptoms={
        "frequent_urination": False,
        "excessive_thirst": False,
        "unexplained_weight_loss": False,
        "fatigue": True,
    },
    fasting_glucose=98,
    random_glucose=145,
)

result_2 = display_result(patient_2)
```

```python
patient_3 = make_patient(
    name="Scenario 3 — Incomplete information",
    symptoms={
        "frequent_urination": True,
        "excessive_thirst": None,
        "unexplained_weight_loss": None,
        "fatigue": True,
    },
    fasting_glucose=None,
    random_glucose=None,
)

result_3 = display_result(patient_3)
```

```python
for patient, result in [
    (patient_1, result_1),
    (patient_2, result_2),
    (patient_3, result_3),
]:
    rule_ids = [rule["id"] for rule in result["activated_rules"]]
    print(f"{patient['name']}: activated rules = {rule_ids or 'None'}")
    print(f"  Missing fields = {len(result['missing_information'])}")

assert "R4" in [rule["id"] for rule in result_1["activated_rules"]]
assert len(result_3["missing_information"]) > 0
print("\nBasic scenario checks passed.")
```
